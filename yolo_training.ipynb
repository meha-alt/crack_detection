```python
from ultralytics import YOLO
```

# Load model

```python
# Using a pretrained model like yolo26n-seg.pt is recommended for faster convergence
model = YOLO("yolo26n-seg.pt")
```

# Train the model on the Crack Segmentation dataset

```python
results = model.train(data="crack-seg.yaml", epochs=100, imgsz=1024)
# trained model will be saved in the 'runs/segment/train/weights as best.pt' by default
```
